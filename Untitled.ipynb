```python
import pandas as pd
from deep_translator import GoogleTranslator
from tqdm import tqdm  # Progress bar library

# Load your CSV
df = pd.read_csv('scientific_subset_3000.csv')

# Initialize Arabic translator
translator = GoogleTranslator(source='auto', target='ar')

# Columns to translate (adjust based on your CSV structure)
text_columns = ['text']  # ← CHANGE THIS to your actual column name(s)

# Translate each row (with error handling)
def safe_translate(text):
    if pd.isna(text) or str(text).strip() == '':
        return text
    try:
        # Skip pure numbers
        if str(text).replace('.', '').replace('-', '').isdigit():
            return text
        return translator.translate(str(text))
    except Exception as e:
        print(f"\nTranslation error: {e}")
        return text

# Apply translation with progress bar
for col in text_columns:
    if col in df.columns:
        print(f"\nTranslating column: '{col}' ({len(df)} rows)")
        tqdm.pandas(desc=f"Translating {col}")
        df[col] = df[col].progress_apply(safe_translate)

# Save with UTF-8 encoding for Arabic support
df.to_csv('scientific_subset_3000_arabic.csv', index=False, encoding='utf-8-sig')
print("\n✅ Translation complete! Saved as 'scientific_subset_3000_arabic.csv'")
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline
from arabert.preprocess import ArabertPreprocessor
import pandas as pd

# Smaller summarization model
model_name = "oddadmix/arabic-summarization"

preprocessor = ArabertPreprocessor(model_name="")
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

summarizer = pipeline("text2text-generation", model=model, tokenizer=tokenizer)

def sum01(text):
    text = preprocessor.preprocess(text)
    result = summarizer(
        text,
        pad_token_id=tokenizer.eos_token_id,
        num_beams=3,
        repetition_penalty=3.0,
        max_length=200,
        length_penalty=1.0,
        no_repeat_ngram_size=3
    )[0]['generated_text']
    return result

# Load dataset and summarize
df = pd.read_csv("/home/kawkab/summerize/datsets/process2/file1.csv").head(2)
df["summary"] = df["text"].apply(sum01)
df.to_csv("/home/kawkab/summerize/datsets/process2/file1.csv", index=False)
```
